### Importing The Required libraries 


```python
import torch
from torch import nn
import torchvision  
from transformers import pipeline 
import cv2
from PIL import Image
import matplotlib.pyplot as plt 
from rembg import remove 
```

### Loading the embedding generation model

```python
embedder=pipeline(model="facebook/dinov2-base",
                  task="image-feature-extraction" )
```

```python
class FinalEmbedder(nn.Module):
    def __init__(self,embedder):
        super().__init__()
        self.embedder=embedder
    def forward(self,image):
        emb=self.embedder(image)
        cls=torch.tensor(emb[0][0])
        return cls
```

```python
def open_image(path):
    img=Image.open(path)
    new=remove(img)
    bbox=new.getbbox()
    img=img.crop(bbox)
    return img
```

### Testing the embedder 

```python
final_embedder=FinalEmbedder(embedder)
```

```python
front_img= open_image(
        "input/Front/inp1.jpeg"
    )
back_img=open_image(
         "input/Back/inp1.jpeg" 
    )
side_img=open_image("input/Side/inp1.jpeg" )
```

```python
front_emb=final_embedder(front_img)
back_emb=final_embedder(back_img)
side_emb=final_embedder(side_img)
```

```python
front_img1=open_image("input/Front/inp2.jpeg")
back_img1=open_image("input/Back/inp2.jpeg")
side_img1=open_image("input/Side/inp2.jpeg")
```

```python
front_emb1=final_embedder(front_img1)
back_emb1=final_embedder(back_img1)
side_emb1=final_embedder(side_img1)
```

### Similarity checking between the embeddings 

```python
def similarity_check(e1,e2):
    similarity=nn.functional.cosine_similarity(e1,e2,dim=0)
    return similarity
```

```python
front_similarity=similarity_check(front_emb,front_emb1)
```

```python
back_similarity=similarity_check(back_emb,back_emb1)
```

```python
side_similarity=similarity_check(side_emb,side_emb1)
```
